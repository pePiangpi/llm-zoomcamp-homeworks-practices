**Import Dependencies and Create RAGTraced**

```python
from opentelemetry import trace
from opentelemetry.sdk.trace import TracerProvider
from opentelemetry.sdk.trace.export import SpanExporter, SpanExportResult, SimpleSpanProcessor
import os
from dotenv import load_dotenv
load_dotenv()

from starter import RAGBase, rag as starter_rag

# 1. Create a clean custom exporter to print span names directly
class SimplePrinter(SpanExporter):
    def export(self, spans):
        for span in spans:
            print(f"-> Span Captured: {span.name}")
        return SpanExportResult.SUCCESS
    def shutdown(self):
        pass

# 2. Setup the Tracer Provider with our clean printer
provider = TracerProvider()
provider.add_span_processor(SimpleSpanProcessor(SimplePrinter()))
trace.set_tracer_provider(provider)
tracer = trace.get_tracer("llm-zoomcamp")

# 3. Define your RAGTraced class
class RAGTraced(RAGBase):
    def __init__(self):
        super().__init__(index=starter_rag.index, llm_client=starter_rag.llm_client)

    def rag(self, query):
        with tracer.start_as_current_span("rag"):
            return super().rag(query)

    def search(self, query):
        with tracer.start_as_current_span("search"):
            return super().search(query)

    def llm(self, prompt):
        with tracer.start_as_current_span("llm"):
            return super().llm(prompt)

# 4. Execute the query
rag = RAGTraced()
query = "How does the agentic loop keep calling the model until it stops?"

print("--- Running Traced RAG Pipeline ---")
answer = rag.rag(query)

print("\n--- Answer ---")
print(answer)
```
